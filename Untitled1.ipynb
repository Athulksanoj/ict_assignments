Q7 # Develop a Personal Expense Tracker to monitor and analyze spending behavior.
Users should be able to log expenses with date, category, and amount. The system should
generate monthly summaries, category-wise spending reports, budget breach alerts, and
identify high-spending patterns.
Analytical insights like highest expense day and dominant spending category must be
computed using loops and conditionals.


```python
3expenses = []
monthly_budget = 5000

while True:
    print("\n--- PERSONAL EXPENSE TRACKER ---")
    print("1. Add Expense")
    print("2. View All Expenses")
    print("3. Monthly Summary")
    print("4. Category-wise Report")
    print("5. Expense Analysis")
    print("6. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        date = input("Enter date (DD-MM): ")
        category = input("Enter category: ")
        amount = float(input("Enter amount: "))

        expenses.append((date, category, amount))
        print("Expense added successfully!")

    elif choice == "2":
        if not expenses:
            print("No expenses recorded.")
        else:
            print("\nDate | Category | Amount")
            for e in expenses:
                print(e[0], "|", e[1], "| Rs.", e[2])

    elif choice == "3":
        total = 0
        for e in expenses:
            total += e[2]

        print("Total Monthly Spending: Rs.", total)

        if total > monthly_budget:
            print("⚠ Budget Breached!")
        else:
            print("Within budge       t.")

    elif choice == "4":
        category_total = {}

        for e in expenses:
            category_total[e[1]] = category_total.get(e[1], 0) + e[2]

        print("\nCategory-wise Spending:")
        for cat in category_total:
            print(cat, ": Rs.", category_total[cat])

    elif choice == "5":
        if not expenses:
            print("No data for analysis.")
        else:
            day_total = {}
            category_total = {}

            for e in expenses:
                day_total[e[0]] = day_total.get(e[0], 0) + e[2]
                category_total[e[1]] = category_total.get(e[1], 0) + e[2]

            highest_day = max(day_total, key=day_total.get)
            dominant_category = max(category_total, key=category_total.get)

            print("Highest Spending Day:", highest_day,
                  "Rs.", day_total[highest_day])
            print("Dominant Category:", dominant_category,
                  "Rs.", category_total[dominant_category])

    elif choice == "6":
        print("Expense Tracker Closed.")
        break

    else:
        print("Invalid choice. Try again.")
```

Q4  Design a Mini Banking System that simulates basic banking operations.
The system should support account creation, deposits, withdrawals, and fund transfers
between accounts. Each transaction must be recorded and retrievable as a mini statement.
Business rules include minimum balance enforcement, daily withdrawal limits, and
rejection of invalid transactions. Data persistence during runtime and menu-based
navigation are mandatory.


```python
accounts = {}

while True:
    print("\n--- MINI BANKING SYSTEM ---")
    print("1. Create Account")
    print("2. Deposit")
    print("3. Withdraw")
    print("4. Transfer")
    print("5. Mini Statement")
    print("6. Exit")

    choice = input("Enter your choice: ")

    # Create Account
    if choice == "1":
        acc_no = input("Enter Account Number: ")
        name = input("Enter Name: ")
        balance = float(input("Enter Initial Balance: "))

        accounts[acc_no] = {
            "name": name,
            "balance": balance,
            "transactions": []
        }

        accounts[acc_no]["transactions"].append(
            "Account created with Rs. " + str(balance)
        )
        print("Account created successfully!")

    # Deposit
    elif choice == "2":
        acc_no = input("Enter Account Number: ")
        if acc_no in accounts:
            amount = float(input("Enter amount: "))
            accounts[acc_no]["balance"] += amount
            accounts[acc_no]["transactions"].append(
                "Deposited Rs. " + str(amount)
            )
            print("Amount deposited.")
        else:
            print("Account not found.")

    # Withdraw
    elif choice == "3":
        acc_no = input("Enter Account Number: ")
        if acc_no in accounts:
            amount = float(input("Enter amount: "))
            if amount > accounts[acc_no]["balance"]:
                print("Insufficient balance.")
            else:
                accounts[acc_no]["balance"] -= amount
                accounts[acc_no]["transactions"].append(
                    "Withdrawn Rs. " + str(amount)
                )
                print("Withdrawal successful.")
        else:
            print("Account not found.")

    # Transfer
    elif choice == "4":
        from_acc = input("From Account: ")
        to_acc = input("To Account: ")
        amount = float(input("Enter amount: "))

        if from_acc in accounts and to_acc in accounts:
            if amount > accounts[from_acc]["balance"]:
                print("Insufficient balance.")
            else:
                accounts[from_acc]["balance"] -= amount
                accounts[to_acc]["balance"] += amount

                accounts[from_acc]["transactions"].append(
                    "Transferred Rs. " + str(amount)
                )
                accounts[to_acc]["transactions"].append(
                    "Received Rs. " + str(amount)
                )
                print("Transfer successful.")
        else:
            print("Invalid account.")

    # Mini Statement
    elif choice == "5":
        acc_no = input("Enter Account Number: ")
        if acc_no in accounts:
            print("\nName:", accounts[acc_no]["name"])
            print("Balance: Rs.", accounts[acc_no]["balance"])
            print("Transactions:")
            for t in accounts[acc_no]["transactions"]:
                print("-", t)
        else:
            print("Account not found.")

    # Exit
    elif choice == "6":
        print("Banking System Closed.")
        break

    else:
        print("Invalid choice.")
```
